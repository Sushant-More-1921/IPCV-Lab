```python
#Edge Detection using canny and sobel algorithm
```

```python
import numpy as np
import cv2
from matplotlib import pyplot as plt
```

```python
#Read image
img = cv2.imread('/content/donaldduck.avif',0)
img
```

```python
# Compute horizontal gradient Gx
sobelx = cv2.Sobel(img,cv2.CV_64F,1,0,ksize=5)
sobelx
```

```python
# Compute vertical gradient Gy
sobely = cv2.Sobel(img,cv2.CV_64F,0,1,ksize=5)
sobely
```

```python
# Compute gradient magnitude
gradmag = np.sqrt(sobelx**2 + sobely**2)
gradmag
```

```python
# Display Sobel edge image
plt.imshow(gradmag,cmap = 'gray')
plt.show()
```

```python
# Canny Edge Detection
img1 = cv2.imread('/content/donaldduck.avif',0)
img1
```

```python
# Apply Gaussian blur
gaussian = cv2.GaussianBlur(img1,(3,3),0)
gaussian
```

```python
# Compute gradient magnitude and direction for Canny
sobelx_canny = cv2.Sobel(gaussian, cv2.CV_64F, 1, 0, ksize=3)
sobely_canny = cv2.Sobel(gaussian, cv2.CV_64F, 0, 1, ksize=3)
mag, angle = cv2.cartToPolar(sobelx_canny, sobely_canny, angleInDegrees=True)
mag
```

```python
# Perform non-maximum suppression
suppressed = np.zeros(img1.shape, dtype=np.uint8)
```

```python
# Apply double threshold
high_threshold = 100
low_threshold = 50
edges = np.zeros(img1.shape, dtype=np.uint8)
edges[mag > high_threshold] = 255
edges[mag > low_threshold] = 128
```

```python
plt.imshow(edges, cmap='gray')
plt.show()
```

```python
# Load the image as a color image for these operations
img_color = cv2.imread('/content/donaldduck.avif')

# Convert BGR image to RGB (for correct display in matplotlib)
img_rgb = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)

# Convert image to grayscale
gray = cv2.cvtColor(img_color, cv2.COLOR_BGR2GRAY)

# Generate Gaussian noise
noise = np.random.normal(0, 25, gray.shape)

# Add noise to grayscale image
noisy = gray + noise
noisy = np.clip(noisy, 0, 255).astype(np.uint8)

# Apply Gaussian Blur (Gaussian Filter)
gaussian_blur = cv2.GaussianBlur(noisy, (5,5), 1)

# Apply Laplacian Filter
log_result = cv2.Laplacian(gaussian_blur, cv2.CV_64F)

# Display all images
plt.figure(figsize=(15,8))

plt.subplot(2,3,5)
plt.imshow(log_result, cmap='gray')
plt.title("LoG Output")
plt.axis('off')
```
